# Задание 1
Написать функцию для определения всех чисел, на которые без остатка делится указанное число.

```python
def find_divisors(n):
    """Функция возвращает список делителей числа n."""
    divisors = []
    for i in range(1, abs(n) + 1):
        if n % i == 0:
            divisors.append(i)
    return divisors

def main_task_1():
    try:
        user_input = input("Введите целое число для поиска делителей: ")
        number = int(user_input)

        result = find_divisors(number)

        print(f"Делители числа {number}: {result}")

    except ValueError:
        print("Ошибка: Введено не целое число! Пожалуйста, используйте только цифры.")
    except Exception as e:
        print(f"Произошла непредвиденная ошибка: {e}")
    finally:
        print("Расчет делителей завершен.")

main_task_1()
```

# Задание 2
Напишите функцию, которая принимает один аргумент и выполняет действия в зависимости от его типа:
- **Список**: Сумма элементов между двумя первыми нулями.
- **Кортеж**: Поменять местами максимальный и минимальный элементы.
- **Число**: Найти сумму четных цифр.
- **Строка**: Перевести каждый символ в код Unicode.

```python
def multi_functional_handler(data):
    print(f"\n--- Обработка типа: {type(data).__name__} ---")
    try:
        if isinstance(data, list):
            first_zero = data.index(0)
            second_zero = data.index(0, first_zero + 1)
            result = sum(data[first_zero + 1 : second_zero])
            print(f"Сумма между нулями: {result}")
            return result

        elif isinstance(data, tuple):
            temp_list = list(data)
            idx_max = temp_list.index(max(temp_list))
            idx_min = temp_list.index(min(temp_list))
            temp_list[idx_max], temp_list[idx_min] = temp_list[idx_min], temp_list[idx_max]
            result = tuple(temp_list)
            print(f"Кортеж после обмена min и max: {result}")
            return result

        elif isinstance(data, (int, float)):
            digits = [int(d) for d in str(abs(int(data))) if d.isdigit()]
            even_sum = sum(d for d in digits if d % 2 == 0)
            print(f"Сумма четных цифр числа: {even_sum}")
            return even_sum

        elif isinstance(data, str):
            unicode_codes = [ord(char) for char in data]
            print(f"Коды Unicode: {unicode_codes}")
            return unicode_codes

        else:
            print("Передан неизвестный тип данных.")

    except ValueError:
        print("Ошибка: В данных нет нужного элемента (например, не хватает нулей в списке) или данные пусты.")
    except Exception as e:
        print(f"Произошла ошибка: {e}")
    finally:
        print("Завершение обработки аргумента.")

try:
    multi_functional_handler([1, 0, 5, 5, 0, 10])
    multi_functional_handler((10, 2, 55, 1, 8))
    multi_functional_handler(123456)
    multi_functional_handler("Python")
except Exception as e:
    print(f"Ошибка при вызове проверок: {e}")
```

# Задание 3
Найти в матрице первую строку, все элементы которой упорядочены по возрастанию.
Изменить упорядоченность элементов этой строки на обратную.

```python
def reverse_first_sorted_row(matrix):
    """Ищет первую возрастающую строку и разворачивает её."""
    found = False

    for i in range(len(matrix)):
        row = matrix[i]
        if row == sorted(row):
            print(f"Найдена подходящая строка на индексе {i}: {row}")
            matrix[i] = row[::-1]
            found = True
            break

    if not found:
        print("В матрице нет строк, упорядоченных по возрастанию.")

    return matrix

def main_task_3():
    print("--- Запуск задания 3 (Матрицы) ---")

    test_matrix = [
        [5, 8, 2],
        [1, 2, 3],
        [4, 5, 6]
    ]

    try:
        print("Исходная матрица:")
        for r in test_matrix: print(r)

        result_matrix = reverse_first_sorted_row(test_matrix)

        print("\nМатрица после обработки:")
        for r in result_matrix: print(r)

    except TypeError:
        print("Ошибка: В матрице должны быть только числа.")
    except Exception as e:
        print(f"Произошла ошибка: {e}")
    finally:
        print("\nРабота с матрицей завершена.")

main_task_3()
```

# Задание 4
Программа, демонстрирующая работу конструкции `try\except\finally`.
Функция деления двух чисел с обработкой ошибок ввода и деления на ноль.

```python
def demonstrate_exception_handling():
    print("--- Демонстрация работы try...except...finally ---")

    try:
        num1 = input("Введите делимое (число): ")
        num2 = input("Введите делитель (число): ")

        result = float(num1) / float(num2)

    except ValueError:
        print("ИСКЛЮЧЕНИЕ: Ошибка значения! Вы ввели не число.")
        result = None

    except ZeroDivisionError:
        print("ИСКЛЮЧЕНИЕ: Ошибка деления на ноль! На 0 делить нельзя.")
        result = None

    except Exception as e:
        print(f"ИСКЛЮЧЕНИЕ: Произошло что-то странное: {e}")
        result = None

    else:
        print(f"УСПЕХ: Результат деления равен {result}")

    finally:
        print("БЛОК FINALLY: Работа функции завершена, ресурсы освобождены.")
        return result

print("Тест 1:")
demonstrate_exception_handling()

print("\nТест 2:")
demonstrate_exception_handling()

print("\nТест 3:")
demonstrate_exception_handling()
```
