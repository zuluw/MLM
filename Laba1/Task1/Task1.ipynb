# Задание 1
Вывести на экран все делители числа. Число вводить с клавиатуры.

```python
try:
    number = int(input("Введите целое число: "))
    print(f"Делители числа {number}:")

    for i in range(1, abs(number) + 1):
        if number % i == 0:
            print(i, end=" ")
except ValueError:
    print("Ошибка! Пожалуйста, введите целое число.")
```

# Задание 2
С клавиатуры вводится строка. Разработать программу, которая реализует указанные действия:
а) подсчитывает количество слов, которые имеют четную длину;
б) выводит на экран самое длинное слово.

```python
user_input = input("Введите предложение: ")

words = user_input.split()

even_count = 0
for word in words:
    clean_word = word.strip(".,!?;:-")
    if len(clean_word) % 2 == 0 and len(clean_word) > 0:
        even_count += 1

if words:
    longest_word = max(words, key=len)

    print(f"Количество слов с четной длиной: {even_count}")
    print(f"Самое длинное слово: {longest_word}")
else:
    print("Вы ввели пустую строку.")
```

# Задание 3
1. Найдите наибольший четный элемент списка. Если такого нет, то выведите первый элемент.
2. Преобразовать список так, чтобы сначала шли положительные элементы, а затем все остальные.

```python
raw_input = input("Введите числа через пробел: ")
my_list = [int(x) for x in raw_input.split()]

if not my_list:
    print("Список пуст!")
else:
    evens = [x for x in my_list if x % 2 == 0]

    if evens:
        result_max = max(evens)
        print(f"Наибольший четный элемент: {result_max}")
    else:
        result_max = my_list[0]
        print(f"Четных элементов нет. Первый элемент списка: {result_max}")

    positive_part = [x for x in my_list if x > 0]
    others_part = [x for x in my_list if x <= 0]

    new_list = positive_part + others_part
    print(f"Преобразованный список: {new_list}")
```

# Задание 4
Найдите три ключа с самыми маленькими значениями в словаре:
`my_dict = {'a':500, 'b':5874, 'c': 560,'d':400, 'e':5874, 'f': 20}`

```python
my_dict = {'a': 500, 'b': 5874, 'c': 560, 'd': 400, 'e': 5874, 'f': 20}

sorted_keys = sorted(my_dict, key=my_dict.get)

result_keys = sorted_keys[:3]

print(f"Исходный словарь: {my_dict}")
print(f"Три ключа с самыми маленькими значениями: {result_keys}")

for key in result_keys:
    print(f"Ключ '{key}', значение: {my_dict[key]}")
```

# Задание 5
Реализуйте программу «Ювелирный магазин».
Словарь: ключ — название, значение — [состав, цена, количество].
Пункты меню:
1. Просмотр описания
2. Просмотр цены
3. Просмотр количества
4. Вся информация
5. Покупка
6. До свидания

```python
items = {
    "Кольцо": ["Золото", 15000, 10],
    "Серьги": ["Серебро", 5000, 20],
    "Подвеска": ["Платина", 25000, 5],
    "Браслет": ["Золото", 12000, 8]
}

while True:
    print("\n--- МЕНЮ МАГАЗИНА ---")
    print("1. Просмотр описания (название – состав)")
    print("2. Просмотр цены (название – цена)")
    print("3. Просмотр количества (название – количество)")
    print("4. Вся информация")
    print("5. Покупка")
    print("6. До свидания")

    choice = input("\nВыберите пункт меню: ")

    if choice == '1':
        for name, info in items.items():
            print(f"{name} – {info[0]}")

    elif choice == '2':
        for name, info in items.items():
            print(f"{name} – {info[1]} руб.")

    elif choice == '3':
        for name, info in items.items():
            print(f"{name} – {info[2]} шт.")

    elif choice == '4':
        print(f"{'Название':<12} | {'Состав':<10} | {'Цена':<8} | {'Кол-во'}")
        print("-" * 45)
        for name, info in items.items():
            print(f"{name:<12} | {info[0]:<10} | {info[1]:<8} | {info[2]}")

    elif choice == '5':
        print("Введите 'n' для выхода из режима покупки.")
        total_sum = 0

        while True:
            buy_item = input("\nВведите название изделия: ")
            if buy_item.lower() == 'n':
                break

            if buy_item in items:
                try:
                    count = int(input(f"Сколько штук '{buy_item}' хотите купить? "))

                    if count <= items[buy_item][2]:
                        cost = count * items[buy_item][1]
                        total_sum += cost

                        items[buy_item][2] -= count

                        print(f"Добавлено в чек. Сумма за {count} шт: {cost}")
                        print(f"Осталось в магазине: {items[buy_item][2]} шт.")
                    else:
                        print("Недостаточно товара на складе!")
                except ValueError:
                    print("Ошибка! Введите числовое количество.")
            else:
                print("Такого товара нет в ассортименте.")

        print(f"\nИТОГО К ОПЛАТЕ: {total_sum} руб.")

    elif choice == '6':
        print("До свидания! Приходите еще.")
        break
    else:
        print("Неверный пункт меню, попробуйте снова.")
```

# Задание 6
Вы принимаете от пользователя последовательность чисел, разделённых запятой. Составьте список и кортеж с этими числами.

```python
user_data = input("Введите числа через запятую (например: 1 2 3 4): ")

numbers_list = [int(x.strip()) for x in user_data.split()]

numbers_tuple = tuple(numbers_list)

print(f"Список (List): {numbers_list}")
print(f"Кортеж (Tuple): {numbers_tuple}")

print(f"Тип первой переменной: {type(numbers_list)}")
print(f"Тип второй переменной: {type(numbers_tuple)}")
```
