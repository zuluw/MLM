# Задание 1
Создать класс `Example`. 
- 2 статические переменные, 2 динамические.
- Метод 1: Создание и вывод локальной переменной.
- Метод 2: Возврат суммы 2-х глобальных переменных.
- Метод 3: Возведение 1-й динамической переменной в степень 2-й.
- Вывод результатов и обращение к переменной `a`.

```python
global_var1 = 10
global_var2 = 5

class Example:
    a = 100
    b = 200

    def __init__(self, dyn1, dyn2):
        self.dyn1 = dyn1
        self.dyn2 = dyn2

    def method1(self):
        local_v = "Я локальная переменная внутри метода 1"
        print(local_v)

    def method2(self):
        return global_var1 + global_var2

    def method3(self):
        return self.dyn1 ** self.dyn2

obj = Example(2, 3)

print("Вызов метода 1:")
obj.method1()

print(f"Результат метода 2 (сумма глобальных): {obj.method2()}")

print(f"Результат метода 3 (возведение в степень): {obj.method3()}")

print(f"Статическая переменная a = {obj.a}")
```

# Задание 2
Создать класс `Employee` с полями: ФИО, стаж, часовая зарплата, количество часов, оклад, премия.
- Реализовать ввод данных с клавиатуры.
- Метод расчета оклада (часовая зарплата * часы).
- Метод расчета премии в зависимости от стажа:
  - до 1 года: 1%
  - до 3 лет: 5%
  - до 5 лет: 8%
  - свыше 5 лет: 15%
- Метод вывода информации.

```python
class Employee:
    def __init__(self, fio, experience, hourly_wage, hours_worked):
        self.fio = fio
        self.experience = experience
        self.hourly_wage = hourly_wage
        self.hours_worked = hours_worked
        self.salary = 0
        self.bonus = 0

    def calculate_salary(self):
        """Рассчитывает оклад на основе отработанных часов."""
        self.salary = self.hourly_wage * self.hours_worked
        return self.salary

    def calculate_bonus(self):
        """Рассчитывает премию в зависимости от стажа."""
        if self.experience < 1:
            percent = 0.01
        elif self.experience < 3:
            percent = 0.05
        elif self.experience < 5:
            percent = 0.08
        else:
            percent = 0.15
        
        self.bonus = self.salary * percent
        return self.bonus

    def display_info(self):
        """Выводит полную информацию о сотруднике."""
        total_pay = self.salary + self.bonus
        print("-" * 30)
        print(f"Сотрудник: {self.fio}")
        print(f"Стаж: {self.experience} лет")
        print(f"Отработано часов: {self.hours_worked}")
        print(f"Оклад: {self.salary:.2f} руб.")
        print(f"Премия: {self.bonus:.2f} руб.")
        print(f"ИТОГО К ВЫПЛАТЕ: {total_pay:.2f} руб.")

employees_list = []

try:
    count = int(input("Сколько сотрудников вы хотите добавить? "))
    
    for i in range(count):
        print(f"\nВвод данных для сотрудника №{i+1}:")
        name = input("Введите ФИО: ")
        exp = float(input("Введите стаж (лет): "))
        wage = float(input("Введите часовую ставку: "))
        hours = float(input("Введите количество отработанных часов: "))
        
        emp = Employee(name, exp, wage, hours)
        
        emp.calculate_salary()
        emp.calculate_bonus()
        
        employees_list.append(emp)

    print("\n" + "="*10 + " СПИСОК СОТРУДНИКОВ " + "="*10)
    for emp in employees_list:
        emp.display_info()

except ValueError:
    print("Ошибка! Пожалуйста, вводите числовые значения для стажа, ставки и часов.")
```

# Задание 3
Создать иерархию классов: Зоомагазин, Животное, Рыбы, Птицы.
- Свойства: порода, стоимость.
- Переопределение метода `movement_method` (способ передвижения).
- Поиск самого дорогого животного.
- Метод записи информации в текстовый файл.

```python
class Animal:
    def __init__(self, breed, price):
        self.breed = breed
        self.price = price

    def movement_method(self):
        return "Передвигается как-то"

    def __str__(self):
        return f"Порода: {self.breed}, Цена: {self.price}, Движение: {self.movement_method()}"

class Fish(Animal):
    def movement_method(self):
        return "Плавает в воде"

class Birds(Animal):
    def movement_method(self):
        return "Летает в небе"

class PetShop:
    def __init__(self, name):
        self.name = name
        self.animals = []

    def add_animal(self, animal):
        self.animals.append(animal)

    def get_most_expensive(self):
        if not self.animals:
            return None
        return max(self.animals, key=lambda a: a.price)

    def save_to_file(self, filename="zoo_inventory.txt"):
        try:
            with open(filename, "w", encoding="utf-8") as f:
                f.write(f"Инвентаризация магазина '{self.name}':\n")
                f.write("-" * 40 + "\n")
                for a in self.animals:
                    f.write(str(a) + "\n")
            print(f"Информация успешно записана в файл: {filename}")
        except Exception as e:
            print(f"Ошибка при записи в файл: {e}")

shop = PetShop("Любимый питомец")

shop.add_animal(Fish("Золотая рыбка", 500))
shop.add_animal(Fish("Акула", 50000))
shop.add_animal(Birds("Волнистый попугай", 1500))
shop.add_animal(Birds("Орёл", 25000))

expensive = shop.get_most_expensive()
if expensive:
    print(f"Самая дорогая порода: {expensive.breed} (Цена: {expensive.price})")

shop.save_to_file()

print("\nСписок всех животных:")
for item in shop.animals:
    print(item)
```

# Задание 4
Реализация класса `Smartphone`, включающая:
1. **Методы экземпляра** (работают с данными конкретного телефона).
2. **Методы класса** (`@classmethod` — работают с самим классом, например, для создания объектов).
3. **Статические методы** (`@staticmethod` — независимые утилиты, логически связанные с классом).

```python
class Smartphone:
    brand_count = 0

    def __init__(self, model, brand, battery=100):
        self.model = model
        self.brand = brand
        self.battery = battery
        Smartphone.brand_count += 1

    def make_call(self, number):
        if self.battery > 5:
            self.battery -= 5
            print(f"Звонок на номер {number} с {self.model}... (Заряд: {self.battery}%)")
        else:
            print("Слишком низкий заряд батареи!")

    @classmethod
    def create_iphone(cls, model):
        print(f"Создаем новый iPhone: {model}")
        return cls(model, "Apple")

    @classmethod
    def get_total_phones(cls):
        return f"Всего выпущено смартфонов: {cls.brand_count}"

    @staticmethod
    def is_valid_number(number):
        return number.startswith('+') and len(number) > 10

print(f"Проверка номера +79991234567: {Smartphone.is_valid_number('+79991234567')}")

phone1 = Smartphone.create_iphone("15 Pro")

phone1.make_call("+79001112233")

phone2 = Smartphone("Galaxy S23", "Samsung", battery=10)
phone2.make_call("+79995554433")

print(Smartphone.get_total_phones())
```
